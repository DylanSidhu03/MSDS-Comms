## 2 · Two favorite visualizations

Some shared sample data:


```python
import pandas as pd
%matplotlib inline

regions = pd.DataFrame({
    'category': ['Alpha', 'Beta', 'Gamma', 'Delta'],
    'North': [420, 310, 260, 190],
    'South': [300, 330, 240, 210],
    'East':  [180, 220, 300, 160],
})

months = pd.DataFrame({
    'month': pd.date_range('2026-01-01', periods=12, freq='MS'),
    'revenue': [100,110,105,130,150,145,160,170,165,180,200,210],
})
```

### Favorite #1 — the grouped bar chart

**Why I like it / the aesthetic choices:**

- **Earthy "Sage, Stone & Clay" palette** — a sage green, a light washed brown, and a
  slate grey lead the palette. It's calm and warm rather than the usual harsh primary colors.
- The colors aren't just picked by eye: I ran them through a **colorblind-safety validator**
  (OKLab ΔE separation) and ordered the slots so neighboring series always stay distinct.
- **Softly rounded bar corners** with an equal *pixel* radius give a modern, friendly feel.
- **Recessive chrome** — no boxed-in axes, just one hairline baseline and faint gridlines,
  plus a warm sand page behind a near-white plot for a little depth. The data stays the star.
- A legend appears **only when there's more than one series**.


```python
sv.bar(regions, 'category', ['North', 'South', 'East'],
       title='Sales by Region',
       subtitle='Earthy palette · rounded bars · recessive chrome',
       ylabel='USD (thousands)')
```

### Favorite #2 — the line chart with area fill

**Why I like it / the aesthetic choices:**

- A single series gets a **soft translucent area fill** under the line for an airy, editorial look.
- **Ringed markers** (a surface-colored edge around each dot) keep points legible on top of the line.
- **Rounded line caps and joins** and a slightly thicker stroke make it feel hand-finished, not default.
- It reuses the same palette and the **title / subtitle / caption** type hierarchy as every other
  chart, so a whole notebook of these reads as one consistent system.
- **Thousands-separated ticks** and muted axis labels keep the numbers quiet and readable.


```python
sv.line(months, 'revenue', x='month',
        title='Monthly Revenue',
        subtitle='Single series · area fill · ringed markers')
```

## 3 · Two problems I ran into (and how I fixed them)


**Problem 1 — the series colors blurred together.**  
Early on I tried very tonal palettes (an all blue-grey set, then an all mint-green set).
They looked nice but adjacent series were nearly impossible to tell apart.  
**Fix:** instead of eyeballing it, I ran candidate palettes through a **colorblind-safety
validator** that scores adjacent-color separation in OKLab ΔE. I spread the hues and lightness
and reordered the palette slots until every neighboring pair cleared the separation threshold —
so the final earthy palette is both pretty *and* distinguishable.


**Problem 2 — rounded bar corners came out lopsided.**  
Matplotlib draws in *data* coordinates, and a bar's x-axis and y-axis have completely different
data-to-pixel scales, so a naive rounded corner looked like a stretched ellipse instead of a circle.  
**Fix:** at draw time I measure the axes' pixel dimensions, compute the x/y pixel ratio, and feed
it into matplotlib's `mutation_aspect` together with a **pixel-based rounding size**. That makes every
corner render as an even quarter-circle no matter the data range or figure size.


**Bonus problem — I accidentally published the wrong package.**  
My first `twine upload` shipped an old `pandaviz 0.1.0` because my local clone was still on the repo's
old default branch, not `main`.  
**Fix:** `git checkout main && git pull`, rebuilt, and republished as the correct
`simpleviz-dgsidhu 0.4.0` — a good reminder to always confirm the branch and built filename before uploading.
